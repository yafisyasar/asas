```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as pt
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
file_path = "//harisree/2024-BCA-C/sathyaprasad.R.Kamath/iris.csv"
data = pd.read_csv(file_path)
print("Original Dataset")
print(data.head())
```

```python
print(data.isnull().sum())
```

```python
encoder = LabelEncoder()
data["species"] = encoder.fit_transform(data["species"])
print("\nEncoded Lable")
print(data.head())
```

```python
x = data.drop("species", axis = 1)
y = data["species"]
print(y)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y , test_size=0.2,random_state=42)
print("\nTraining Sample ",len(x_train))
print("\nTest Sample ",len(x_test))
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
print(x_train_scaled[:5])
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(x_train_scaled, y_train)
y_pred = knn.predict(x_test_scaled)
```

```python
print("Accuracy : ",accuracy_score(y_test,y_pred))
print("Confusion Matrix :\n",confusion_matrix(y_test,y_pred))
print("ClassificationReport :\n",classification_report(y_test,y_pred))
```

```python
new = [[1.3, 2.1, 1.2, 0.3]]
new_scaled = scaler.transform(new)
pred = knn.predict(new_scaled)
print("Prediction : ",pred)
```
