# Sales prediction using Advertisement through TV

```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
```

```python
# Read the given CSV file, and view some sample records
advertising = pd.read_csv("Z:/Machine_learning Dr. Vimina/Machine Learning 2026/BCA H 2026/ML Lab Programs/linearregression code/Company_data.csv")
advertising
```

```python
# Info our dataset
advertising.info()
```

```python
# Describe our dataset
advertising.describe()
```

```python
# Creating X and y for demonstrating Simple Linear Regression
X = advertising[["TV"]]
y = advertising["Sales"]
```

```python
# Splitting the varaibles as training and testing
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size = 0.7,test_size=0.3,random_state=100)
```

```python
print(X_train)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.fit_transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
slr = LinearRegression()
slr.fit(X_train_scaled, y_train)
#Displaying Intercept and coefficients
print('Intercept: \n', slr.intercept_)
print('Coefficients: \n', slr.coef_)
```

```python
# Make predictions using the testing set

y_pred = slr.predict(X_test_scaled)
```

```python
#Model Evaluation
print("MAE :", round(mean_absolute_error(y_test, y_pred), 2))
print("MSE :", round(mean_squared_error(y_test, y_pred), 2))
print("RMSE:", round(np.sqrt(mean_squared_error(y_test, y_pred)), 2))
print("R2 Score:", round(r2_score(y_test, y_pred), 2))
```

```python
# Plot outputs
plt.scatter(X_test_scaled, y_test, color="black")
plt.plot(X_test_scaled, y_pred, color="blue", linewidth=3)

plt.xticks(())
plt.yticks(())

plt.show()
```
