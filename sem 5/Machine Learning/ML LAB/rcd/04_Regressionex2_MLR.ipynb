Multiple Linear Regression

```python
# Creating X and y
X = advertising[['TV', 'Radio','Newspaper']]
y = advertising['Sales']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size = 0.7,test_size=0.3,random_state=100)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.fit_transform(X_test)
```

```python
mlr = LinearRegression()
mlr.fit(X_train_scaled, y_train)
```

```python
# Display the y intercept and the feature coefficients
print("Intercept: ", mlr.intercept_)
print("Coefficients:")
list(zip(X, mlr.coef_))
```

```python
#Prediction of test set
y_pred_mlr= mlr.predict(X_test)
#Predicted values
print("Prediction for test set: {}".format(y_pred_mlr))
```

```python
#Actual value and the predicted value
mlr_diff = pd.DataFrame({'Actual value': y_test, 'Predicted value': y_pred_mlr})
mlr_diff.head()
```

```python
#Model Evaluation
print("MAE :", round(mean_absolute_error(y_test, y_pred_mlr), 2))
print("MSE :", round(mean_squared_error(y_test, y_pred_mlr), 2))
print("RMSE:", round(np.sqrt(mean_squared_error(y_test, y_pred_mlr)), 2))
print("R2 Score:", round(r2_score(y_test, y_pred_mlr), 2))
```
