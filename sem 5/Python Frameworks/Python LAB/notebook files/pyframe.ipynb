# Analysing a Dataset Using NumPy​

```python
marks = np.array([
[80, 75, 85],
[65, 70, 60],
[90, 88, 95],
[72, 80, 78],
[85, 92, 88]
])
print("Student Marks:")
print(marks)
```

```python
# Basic properties​
print("\nNumber of dimensions:", marks.ndim)
print("Shape:", marks.shape)
print("Number of elements:", marks.size)
```

```python
# Total marks for each student​
total = np.sum(marks, axis=1)
print("\nTotal marks:")
print(total)
```

```python
# Average marks for each student​
average = np.mean(marks, axis=1)
print("\nAverage marks:")
print(average)
```

```python
# Subject-wise average​
subject_average = np.mean(marks, axis=0)
print("\nSubject-wise average:")
print(subject_average)
```

```python
# Highest and lowest marks​
print("\nHighest mark:", np.max(marks))
print("Lowest mark:", np.min(marks))
```

```python
# Highest mark in each subject​
print("\nHighest mark in each subject:")
print(np.max(marks, axis=0))
```

```python
# Lowest mark in each subject​
print("\nLowest mark in each subject:")
print(np.min(marks, axis=0))
```

```python
# Students with average above 80
print("\nAverages above 80:")
print(average[average > 80])
```

```python
# Sorted averages​
print("\nSorted averages:")
print(np.sort(average))
```
