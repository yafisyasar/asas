```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "Name": ["Anu", "Binu", "Cathy", "David"],
    "Age": [20, np.nan, 19, 22],
    "Marks": [85, 90, np.nan, 88],
    "City": ["Kochi", "Kochi", np.nan, "Thrissur"]
})
```

```python
print("Original dataset:")
print(df)
```

```python
print("\nMissing values:")
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
```

```python
df["City"] = df["City"].fillna(df["City"].mode()[0])
```

```python
print("\nCleaned dataset:")
print(df)
```
